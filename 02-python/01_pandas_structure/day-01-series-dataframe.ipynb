Explicação - explanation

1. DataFrame is a complete table with rows and columns. Think of it as an Excel spreadsheet or a SQL table.



2. A Series is a single column of this table, with an index on the side. When you do df["value"], what you get is a Series.  

3. An Index is a label of each row. By default it's 0, 1, 2, 3... but it can be something else.

4. Before analyzing any table, we inspect it. The five main commands for this are:

- df.shape: returns (rows, columns). It's an attribute, so it doesn't have parentheses.
- df.dtypes: shows the type of each column (int64, float64, object for text etc.).
- df.info (): general summary with columns, types and number of non-null values.
- df.describe(): statistics for numerical columns (mean, minimum, maximum, quartiles)
- df.head (n) and df.tail (n): first and last n rows.

```python
# Example
# You can create a dataframe directly from a dictionary, without needing a CSV.
# Each keys becomes a column.

import pandas as pd

dados = {
    "produto": ["camiseta", "calça", "boné", "tênis"],
    "qtde": [10, 4, 7, 2],
    "preco": [49.9, 119.9, 35.0, 249.9],
}

df = pd.DataFrame(dados)

print(df.shape)
print("----------------------------------")
print(df.dtypes)
print("----------------------------------")
print(df.info())
print("----------------------------------")
print(df.describe())
print("----------------------------------")
print(df.head(2))
print("----------------------------------")
print(df.tail(2))
```

Note that df["preco"] returns a Series, and Series have their own methods, such as .mean(), .sum(), and .max().

```python
# Exercises

import pandas as pd

dados = {
    "pedido": [101, 102, 103, 104, 105, 106],
    "cliente": ["Ana", "Bruno", "Carla", "Ana", "Diego", "Bruno"],
    "produto": ["camiseta", "tênis", "boné", "calça", "camiseta", "boné"],
    "qtd": [2, 1, 3, 1, 5, 2],
    "preco_unit": [49.9, 249.9, 35.0, 119.9, 49.9, 35.0],
}

df = pd.DataFrame(dados)
```
