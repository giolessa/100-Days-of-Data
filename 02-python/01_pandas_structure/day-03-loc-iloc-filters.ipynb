# Explanation

1. loc select for label (the index name and the column name). iloc select for position (0, 1, 2...), such as in python lists.
2. The syntax of both: df.loc[rows, columns] and df.iloc[rows, columns]. If you omit the columns, all of them are returned.
3. Slicing has an important difference:

    - iloc[0:3] gets positions 0, 1, and 2, meaning the stop bound is exclusive, just like Python lists.

    - loc[0:3] gets labels 0, 1, 2, and 3, meaning the stop bound is inclusive.

4. Conditional filtering returns a Series of Booleans (True/False), which is called a mask. When you pass this mask into df[...], only the True rows are kept. This is equivalent to SQL's WHERE.
5. Multiple conditions: use & (and), | (or), and ~ (not). Each condition must be wrapped in parentheses because & and | have higher operator precedence than comparison operators like > and ==. Also, Python's built-in and and or keywords cannot be used here.
6. Filter shortcuts: col.isin([a, b]) acts like SQL's IN.
7. col.between(a, b) acts like BETWEEN, and it includes both endpoints (inclusive).
8. Filtering rows and columns at the same time: df.loc[mask, ["col1", "col2"]] filters the rows and selects the columns in a single line. It's like combining WHERE and SELECT together.

# EXEMPLE

```python
import pandas as pd

filmes = pd.DataFrame ({
    "titulo": ["A", "B", "C", "D", "E"],
    "genero": ["drama", "ação", "drama", "comédia", "ação"],
    "nota": [8.1, 6.5, 7.2, 5.9, 8.8],
    "duracao": [120, 95, 140, 88, 150]
})

filmes.loc[2]                       # the row index 2
```

```python
filmes.iloc[0:3]                    # positions 0, 1 e 2
```

```python
filmes.loc[0:3]                     # labels 0, 1, 2 e 3 (4 rows!)
```

```python
filmes.loc[:, ["titulo", "nota"]]   # all rows, 2 columns
```

```python
filmes.iloc[1:4, 0:2]               # rows 1 to 3, columns 0 to 1
```

```python
filmes[filmes["nota"] > 7]                                   # A filter
```

```python
filmes[(filmes["nota"] > 7) & (filmes["duracao"] < 145)]     # AND
```

```python
filmes[(filmes["genero"] == "ação") | (filmes["nota"] > 8)]  # OR
```

```python
filmes[filmes["genero"].isin(["drama", "comédia"])]          # IN
```

```python
filmes[filmes["nota"].between(6, 8)]                         # BETWEEN
```

```python
filmes[~filmes["genero"].isin(["ação"])]                     # NOT
```

```python
filmes.loc[filmes["nota"] > 7, ["titulo", "genero"]]         # filter + columns
```

```python
# EXERCICIO

import pandas as pd

dados = {
    "pedido": [101, 102, 103, 104, 105, 106, 107, 108],
    "cliente": ["Ana", "Bruno", "Carla", "Ana", "Diego", "Bruno", "Elisa", "Diego"],
    "cidade": ["Rio", "Niterói", "Rio", "Rio", "São Gonçalo", "Niterói", "Rio", "São Gonçalo"],
    "produto": ["camiseta", "tênis", "boné", "calça", "camiseta", "boné", "tênis", "calça"],
    "qtd": [2, 1, 3, 1, 5, 2, 1, 2],
    "preco_unit": [49.9, 249.9, 35.0, 119.9, 49.9, 35.0, 249.9, 119.9],
}

df = pd.DataFrame(dados)
df["faturamento"] = df["qtd"] * df["preco_unit"]
```

```python
# Show the row index 4 (all columns)

df.loc[4]
```

```python
# Using iloc, show the first 3 rows and only the 'cliente' and 'produto' columns. Think about the position of the columns in the DataFrame.

df.iloc[:3, [1, 3]]
```

```python
# Show the orders with qtd greater than or equal to 3.

df[df["qtd"] >= 3]
```

```python
# Show the orders with qtd greater than or equal to 2 AND preco_unit less than 100.

df[(df["qtd"] >= 2) & (df["preco_unit"] < 100)]
```

```python
# Show the orders from customers Ana or Diego, using isin.

df[df["cliente"].isin(["Ana", "Diego"])]
```

```python
# Show the orders with preco_unit between 40 and 120 (inclusive), using between.

df[df["preco_unit"].between(40, 120)]
```

```python
# Show only the cliente and produto columns for orders with faturamento greater than 100, in a single line using loc.

df.loc[df["faturamento"] > 100, ["cliente", "produto"]]
```

```python
# BONUS
# The orders that are not for a t-shirt and are not from the city of Rio.
# Think carefully about the parentheses and which operator to use.

df[(~df["cidade"].isin(["Rio"])) & (~df["produto"].isin(["camiseta"]))]
```
