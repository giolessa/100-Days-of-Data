1. What is a NumPy array and why use it?

NumPy is the foundation of almost all data analysis in Python, and Pandas is built on top of it. Its core object is the ndarray (n-dimensional array): a "list" optimized for numbers.

A Python list can store anything mixed together (numbers, strings, other lists), and each element is a separate object in memory. A NumPy array stores elements of the same data type, contiguously in memory. This makes it much faster and allows you to perform calculations across the entire collection at once.

```python
import numpy as np
```

# Creating arrays

```python
# From a list

a = np.array([28, 2, 2007])
print(a)
print(type(a))
```

```python
# With built-in functions

np.zeros(3)          # [0. 0. 0.]
np.ones((2, 3))      # [[1. 1. 1.]
                     #  [1. 1. 1.]]

np.arange(0, 10, 2)  # [0 2 4 6 8]
np.linspace(0, 1, 5) # [0.   0.25 0.5  0.75 1.  ]

# arange works like Python's range, but returns an array.
# linspace means "give me N evenly spaced points between A and B (inclusive)"
```

# shape, ndim e size: the "array" format

```python
a = np.array([10, 20, 30])
b = np.array([[1, 2, 3],
              [4, 5, 6]])

print(a.shape)   # (3,)    -> 1 dimension, 3 elements
print(b.shape)   # (2, 3)  -> 2 rows, 3 columns

print(a.ndim)    # 1       -> number of dimensions
print(b.ndim)    # 2

print(b.size)    # 6       -> total elements (2 x 3)
```

dtype: the type of data

```python
# All elements in an array have the same data type, which is stored in `dtype`.

np.array([1, 2, 3]).dtype        # int64   (integers)
np.array([1.5, 2.0, 3.1]).dtype  # float64 (floats / decimals)

# If you mix types, NumPy converts everything to the most general type:

np.array([1, 2.5, 3])
# [1.  2.5 3. ]  -> the int became a float (dtype float64)

np.array([1, "a", 3])
# ['1' 'a' '3']  -> everything became strings (dtype '<U21')

# You can convert the type using `astype`:

x = np.array([1.7, 2.2, 3.9])
print(x.astype(int))   # [1 2 3]  -> note: it TRUNCATES, it does not round
```

# Vectorized operations: the superpower

```python
# "Vectorizing" means applying an operation to all elements at once, without writing a for loop.

# Array with a number (scalar):

a = np.array([1, 2, 3])
print(a * 2)    # [2 4 6]
print(a + 10)   # [11 12 13]
print(a ** 2)   # [1 4 9]
```

```python
# Array with array (element-wise):

x = np.array([10, 20, 30])
y = np.array([1, 2, 3])

print(x + y)    # [11 22 33]
print(x * y)    # [10 40 90]
print(x / y)    # [10. 10. 10.]
```

```python
# Comparisons are also vectorized and return an array of True/False. This is the foundation of the filters you've already seen in Pandas:

print(x > 15)   # [False  True  True]
```

```python
# Comparing with a regular list:

[1, 2, 3] * 2             # [1, 2, 3, 1, 2, 3]  -> list REPEATS
np.array([1, 2, 3]) * 2   # [2 4 6]             -> array MULTIPLIES
```
